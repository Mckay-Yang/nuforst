# Visualize Region of Interest
Show a 1024x1024 pixel region around a specific coordinate using geemap.

```python
# %pip install geemap
```

```python
import geemap
import ee

# Authenticate and Initialize Earth Engine
try:
    ee.Initialize(project='ee-yangluhao990714')
except Exception:
    ee.Authenticate()
    ee.Initialize(project='ee-yangluhao990714')

# Parameters
coord = (94.2604595275288, 29.773270070797224) # (Longitude, Latitude)
pixel_size = 1024
scale = 10  # meters per pixel

# Calculate Bounding Box
# 1024 pixels * 10m = 10240m width
# Buffer radius = width / 2
radius = (pixel_size * scale) / 2

point = ee.Geometry.Point(list(coord))
# Use buffer().bounds() to get a square bounding box aligned with axes
bbox = point.buffer(radius).bounds()

# Visualization
Map = geemap.Map()
Map.centerObject(point, 13)

# Add Satellite Basemap for context
Map.add_basemap('HYBRID')

# Add layers
Map.addLayer(bbox, {'color': 'blue'}, '1024x1024 Area', opacity=0.3)
Map.addLayer(point, {'color': 'red'}, 'Center Point')

# Print info
print(f"Center: Lon {coord[0]}, Lat {coord[1]}")
print(f"Scale: {scale}m/px")
print(f"Region Size: {pixel_size}x{pixel_size} pixels ({pixel_size*scale}m x {pixel_size*scale}m)")

Map
```
