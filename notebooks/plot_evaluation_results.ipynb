# Evaluation Results Visualization

This notebook reads the evaluation results (`data/output/evaluation_results.csv`) and generates bar charts comparing the performance of different methods (`NuFrost`, `Zhu2015`, `HANTS`) across different Sentinel-2 bands.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from pathlib import Path

# Set plotting style suitable for scientific papers
sns.set_theme(style="whitegrid", font_scale=1.2, rc={"axes.edgecolor": "black", "axes.linewidth": 1})

# Define a clear, high-contrast palette
# NuFrost: Red, Zhu2015: Blue, HANTS: Green
method_palette = {'NuFrost': '#e44d58', 'Zhu2015': '#4284f5', 'HANTS': '#fbbc02'}

# Load the data
data_path = '../data/output/evaluation_results_hls.csv'
if Path(data_path).exists():
    df = pd.read_csv(data_path)

    # Extract the band from the 'Image' filename (e.g., GREEN_lon118.1656_lat30.1375 -> GREEN)
    df['Band'] = df['Image'].str.extract(r'^([A-Za-z0-9]+)_lon')

    # Define standard optical band order for sorting
    band_order = ['BLUE', 'GREEN', 'RED', 'NIR1', 'SWIR1', 'SWIR2']
    # Create a categorical type with the specified order to sort the dataframe
    df['Band'] = pd.Categorical(df['Band'], categories=band_order, ordered=True)
    df = df.dropna(subset=['Band'])
    df = df.sort_values('Band')

    print("Data loaded successfully. Preview:")
    display(df.head())
else:
    print(f"File not found: {data_path}")
```

## 1. Root Mean Square Error (RMSE)
Lower RMSE values indicate better performance.

```python
plt.figure(figsize=(10, 6))
ax = sns.barplot(
    data=df,
    x='Band',
    y='RMSE',
    hue='Algorithm',
    palette=method_palette,
    capsize=.1,
    err_kws={'linewidth': 1.5, 'color': 'black'}
)

plt.title('Root Mean Square Error (RMSE) by Band', pad=15, fontweight='bold')
plt.xlabel('Sentinel-2 Band', fontweight='bold')
plt.ylabel('RMSE', fontweight='bold')
plt.legend(title='Method', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## 2. Mean Absolute Error (MAE)
Lower MAE values indicate better performance.

```python
plt.figure(figsize=(10, 6))
ax = sns.barplot(
    data=df,
    x='Band',
    y='MAE',
    hue='Algorithm',
    palette=method_palette,
    capsize=.1,
    err_kws={'linewidth': 1.5, 'color': 'black'}
)

plt.title('Mean Absolute Error (MAE) by Band', pad=15, fontweight='bold')
plt.xlabel('Sentinel-2 Band', fontweight='bold')
plt.ylabel('MAE', fontweight='bold')
plt.legend(title='Method', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## 3. Correlation Coefficient (R)
Higher correlation coefficient values (closer to 1.0) indicate better performance.

```python
plt.figure(figsize=(10, 6))
ax = sns.barplot(
    data=df,
    x='Band',
    y='R',
    hue='Algorithm',
    palette=method_palette,
    capsize=.1,
    err_kws={'linewidth': 1.5, 'color': 'black'}
)

plt.title('Pearson Correlation Coefficient (R) by Band', pad=15, fontweight='bold')
plt.xlabel('Sentinel-2 Band', fontweight='bold')
plt.ylabel('R', fontweight='bold')
plt.legend(title='Method', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
